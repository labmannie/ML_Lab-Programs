```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error

data = {
    'x': [1, 2, 3, 4, 5, 6, 7, 8, 9],
    'y': [5, 7, 9, 11, 50, 15, 17, 19, 21]
}

df = pd.DataFrame(data)

train_df = df[df['x'].between(1, 5)].copy()
val_df   = df[df['x'].between(6, 7)].copy()
test_df  = df[df['x'].between(8, 9)].copy()

X_train, y_train = train_df[['x']], train_df['y']
X_val,   y_val   = val_df[['x']],   val_df['y']
X_test,  y_test  = test_df[['x']],  test_df['y']

models = {
    'OLS': LinearRegression(),
    'Ridge (alpha=1.0)': Ridge(alpha=1.0),
    'Lasso (alpha=1.0)': Lasso(alpha=1.0)
}

def generate_prediction_table(model, X, y):
    preds = model.predict(X)
    errors = y - preds
    sq_errors = errors ** 2

    return pd.DataFrame({
        'x': X['x'].values,
        'Actual y': y.values,
        'Predicted y_hat': np.round(preds, 3),
        'Error (y - y_hat)': np.round(errors, 3),
        'Squared Error (e^2)': np.round(sq_errors, 4)
    })

for name, model in models.items():
    model.fit(X_train, y_train)

    slope = model.coef_[0]
    intercept = model.intercept_

    train_mse = mean_squared_error(y_train, model.predict(X_train))
    val_mse   = mean_squared_error(y_val,   model.predict(X_val))
    test_mse  = mean_squared_error(y_test,  model.predict(X_test))

    print("=" * 60)
    print(f"MODEL: {name}")
    print(f"Equation: y_hat = {slope:.3f}x + {intercept:.3f}")
    print("=" * 60)

    for split_label, X_data, y_data, mse_val in [('TRAINING (x=1 to 5)', X_train, y_train, train_mse),
                                                ('VALIDATION (x=6 to 7)', X_val, y_val, val_mse),
                                                ('TESTING (x=8 to 9)', X_test, y_test, test_mse)]:
        print(f"\n--- {split_label} ---")
        # Removed detailed prediction table for brevity as requested
        # pred_table = generate_prediction_table(model, X_data, y_data)
        # print(pred_table.to_string(index=False))
        print(f"MSE: {mse_val:.4f}")

    print("\n")
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error

# Dataset containing the outlier at x=5, y=50
data = {
    'x': [1, 2, 3, 4, 5, 6, 7, 8, 9],
    'y': [5, 7, 9, 11, 50, 15, 17, 19, 21]
}

df = pd.DataFrame(data)

# Splitting data according to Case B requirements
train_df = df[df['x'].between(1, 5)].copy()
val_df   = df[df['x'].between(6, 7)].copy()
test_df  = df[df['x'].between(8, 9)].copy()

X_train, y_train = train_df[['x']], train_df['y']
X_val,   y_val   = val_df[['x']],   val_df['y']
X_test,  y_test  = test_df[['x']],  test_df['y']

# Models with lambda/alpha = 10 for regularization
models = {
    'OLS': LinearRegression(),
    'Ridge (alpha=10)': Ridge(alpha=10.0),
    'Lasso (alpha=10)': Lasso(alpha=10.0)
}

for name, model in models.items():
    model.fit(X_train, y_train)

    slope = model.coef_[0]
    intercept = model.intercept_

    train_mse = mean_squared_error(y_train, model.predict(X_train))
    val_mse   = mean_squared_error(y_val,   model.predict(X_val))
    test_mse  = mean_squared_error(y_test,  model.predict(X_test))

    print("=" * 60)
    print(f"MODEL: {name}")
    print(f"Equation: y_hat = {slope:.3f}x + {intercept:.3f}")
    print("=" * 60)
    print(f"Train MSE: {train_mse:.4f}")
    print(f"Val MSE:   {val_mse:.4f}")
    print(f"Test MSE:  {test_mse:.4f}\n")
```

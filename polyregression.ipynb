```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

X2 = np.array([[-2], [-1], [0], [1], [2]])
y2 = np.array([9, 2, 1, 6, 17])

poly2 = PolynomialFeatures(degree=2)
X2_poly = poly2.fit_transform(X2)

model2 = LinearRegression(fit_intercept=False)
model2.fit(X2_poly, y2)
pred2 = model2.predict(X2_poly)

print("Degree 2 Coefficients:", np.round(model2.coef_[::-1], 2))
print("Degree 2 Intercept:", model2.coef_[0], 2)
print("Degree 2 MSE:", round(mean_squared_error(y2, pred2), 2))
print("Degree 2 R2 Score:", round(model2.score(X2_poly, y2), 2))

X3 = np.array([[-2], [-1], [0], [1], [2]])
y3 = np.array([-11, -2, 1, 4, 13])

poly3 = PolynomialFeatures(degree=3)
X3_poly = poly3.fit_transform(X3)

model3 = LinearRegression(fit_intercept=False)
model3.fit(X3_poly, y3)
pred3 = model3.predict(X3_poly)

print("\nDegree 3 Coefficients:", np.round(model3.coef_[::-1], 2))
print("Degree 3 Intercept:", model3.coef_[0], 2)
print("Degree 3 MSE:", round(mean_squared_error(y3, pred3), 2))
print("Degree 3 R2 Score:", round(model3.score(X3_poly, y3), 2))
```
